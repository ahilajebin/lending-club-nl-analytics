Load the cleaned dataset

```python
import pandas as pd

df = pd.read_csv("../data/processed/clean_lending_club.csv")

print(df.shape)
df.head()
```

Final list of features/columns selected to load into BigQuery

```python
final_cols = [
    "loan_amnt",
    "int_rate",
    "annual_inc",
    "emp_length",
    "grade",
    "home_ownership",
    "purpose",
    "loan_status",
    "issue_d",
    "dti"
    
]

df_final = df[final_cols]
```

```python
df_final.info()
df_final.isnull().sum()
```

Saving this as the final dataset

```python
import os

os.makedirs("../data/processed", exist_ok=True)

df_final.to_csv(
    "../data/processed/clean_lending_club.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
print("Shape:", df_final.shape)
```
