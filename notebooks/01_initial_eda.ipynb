```python
import pandas as pd
df = pd.read_csv("../data/raw/loan.csv", low_memory=False)
#sample
df = df.sample(n=100_000, random_state=42)
df.shape
```

```python
df.info()
```

```python
df.head()
```

```python
#Missing Values
missing_val = (df.isnull().mean().sort_values(ascending=False))
missing_val.head(20)
```

```python
missing_val.shape[0]
```

```python
acceptable1 = missing_val[missing_val<=0.25].index
acceptable1.shape[0]
```

```python
acceptable1
```

```python
#Cardinality check to find high-cardinality categorical columns
cardinality = df.nunique().sort_values(ascending=False)
cardinality.head(20)
```

```python
cardinality.shape[0]
```

```python
acceptable2 = cardinality[cardinality<=100].index
```

```python
acceptable2
```

```python
acceptable2.shape[0]
```

```python
# Choosing the below columns from acceptable1 and acceptable2 list:
# loan_amt, int_rate, annual_inc, emp_length, grade, home_ownership, 
# purpose, loan_status, issue_d
final_data = df[["loan_amnt", "int_rate", "annual_inc", "emp_length", 
                 "grade", "home_ownership", "purpose", "loan_status", "issue_d"]]
```

```python
final_data.info()
```

```python
final_data.isnull().sum()
```

```python
# Saving cleaned data file as csv
final_data.to_csv("../data/processes/clean_lending_club.csv", index=False)
```
